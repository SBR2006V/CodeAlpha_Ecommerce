```python
# Program 6
with open("student.txt","r") as file:
  text = file.read()
vowels = 0
consonants = 0
digits = 0
special = 0
for ch in text:
 if ch.lower() in "aeiou":
  vowels += 1
 elif ch.isalpha():
  consonants += 1
 elif ch.isdigit():
  digits += 1
 elif not ch.isspace():
  special += 1
print("Vowels =", vowels)
print("Consonants =", consonants)
print("Digits =", digits)
print("Special Characters =", special)
```

```python
# Program 7
word = input("Enter word to search: ")
with open("student.txt","r") as file:
 text = file.read()
count = text.lower().count(word.lower())
if count > 0:
 print(word,"found",count,"time(s).")
else:
 print(word,"not found.")
```

```python
# Program 8
old_word = input("Enter word to replace: ")
new_word = input("Enter new word: ")
with open("student.txt","r") as file:
 text = file.read()
updated_text = text.replace(old_word,new_word)
with open("student.txt","w") as file:
 file.write(updated_text)
print("\nUpdated File:\n")
with open("student.txt","r") as file:
 print(file.read())
```

```python
# Program 10
while True:
    print("\n===== Student Record Management =====")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Append Student")
    print("5. Exit")
    choice = input("Enter Choice: ")

    if choice == "1":
        name = input("Enter Name: ")
        roll = input("Enter Roll: ")
        branch = input("Enter Branch : ")
        marks = input("Enter Marks: ")
        with open("students.txt", "w") as file:
            file.write(name + "," + roll + "," + branch + ", " + marks + "\n")
        print("Student Record Added.")

    elif choice == "2":
        try:
            with open("students.txt", "r") as file:
                print("\nStudent Records\n")
                print(file.read())
        except FileNotFoundError:
            print("No records found.")

    elif choice == "3":
        roll = input("Enter Roll Number: ")
        found = False
        try:
            with open("students.txt", "r") as file:
                for line in file:
                    record = line.strip().split(",")
                    if record[1] == roll:
                        print("\nRecord Found")
                        print("Name:", record[0])
                        print("Roll :", record[1])
                        print("Branch:", record[2])
                        print("Marks:", record[3])
                        found = True
            if not found:
                print("Student Not Found")
        except FileNotFoundError:
            print("File does not exist.")

    elif choice == "4":
        name = input("Enter Name: ")
        roll = input("Enter Roll: ")
        branch = input("Enter Branch: ")
        marks = input("Enter Marks: ")
        with open("students.txt", "a") as file:
            file.write(name + ", " + roll + "," + branch + ", " + marks + "\n")
        print("Student Record Appended.")

    elif choice == "5":
        print("Program Ended.")
        break

    else:
        print("Invalid Choice.")
```
