```python
text = str(input("Enter a string: "))
vowels = 0
consonants = 0
for ch in text:
 if ch.isalpha():
  if ch in "aeiou":
    vowels += 1
  else:
    consonants += 1
print("Vowels =", vowels)
print("Consonants =", consonants)
```

```python

text = str(input("Enter a string: "))
upper = 0
lower = 0
for ch in text:
 if ch.isupper():
  upper += 1
 elif ch.islower():
  lower += 1
print("Uppercase Letters =", upper)
print("Lowercase Letters =", lower)
```

```python
text = str(input("Enter a string: "))
frequency = {}
for ch in text:
  frequency[ch] = frequency.get(ch, 0) + 1
print("Charecter Frequency: ")
for key, value in frequency.items():
  print(key, ":", value)
```
